# 🔍 EcoQuery Vocabulary Gap Demo

## Bridging the Gap Between User Language and Environmental Database Terminology

**The Problem:** Users search for generic terms like "steel" but environmental databases contain specific products like "steel, low-alloyed", "steel, chromium steel 18/8", etc.

**The Solution:** Use semantic search with NLP to automatically suggest specific database products for generic user queries.

---

**Dataset:** 4,295 products from ecoinvent v3.11 intermediate exchanges  
**Method:** Sentence transformers for semantic similarity matching  
**Evaluation:** 25 test queries across different categories

```python
# Force complete reload - more aggressive
import sys
import os
import importlib

# Get paths
notebook_dir = os.getcwd()
project_dir = os.path.dirname(notebook_dir)
src_dir = os.path.join(project_dir, 'src')

# Add to path
if src_dir not in sys.path:
    sys.path.append(src_dir)

# Nuclear option - remove ALL references
for module in list(sys.modules.keys()):
    if 'semantic_matching' in module:
        del sys.modules[module]

# Fresh import
from semantic_matching import ProductMatcher
import pandas as pd

print("🚀 Setting up semantic search system...")
print(f"Using src directory: {src_dir}")
```

```python
# Initialize the semantic search system
print("Loading sentence transformer model...")
matcher = ProductMatcher()

print("Loading product database...")
matcher.load_data()

print("Creating semantic embeddings...")
matcher.create_embeddings()

print("✅ System ready! Let's see the magic...")
```

```python
print("SEMANTIC SEARCH DEMO")
print("=" * 50)
print("Regional terminology mapping: US -> UK")
print()

query = "gasoline"
matches = matcher.find_matches(query, top_k=3)

print(f"User searches: '{query}'")
for i, match in enumerate(matches, 1):
    print(f"  {i}. {match['product_name']} ({match['similarity']:.3f})")
```
