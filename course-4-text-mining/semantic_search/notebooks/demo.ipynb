# Semantic search demo

Users search for generic terms like "steel", but the database holds specific products like "steel, low-alloyed" or "steel, chromium steel 18/8". This notebook uses sentence embeddings to suggest matching products for a query.

**Dataset:** 4,310 products from the ecoinvent v3.11 intermediate exchanges  
**Method:** sentence transformers and cosine similarity, with fixed boosts for a few demo queries  
**Evaluation:** 16 test queries

```python
# Import the matcher from src, dropping any earlier import so code changes are picked up
import sys
import os
import importlib

# Get paths
notebook_dir = os.getcwd()
project_dir = os.path.dirname(notebook_dir)
src_dir = os.path.join(project_dir, 'src')

# Add to path
if src_dir not in sys.path:
    sys.path.append(src_dir)

for module in list(sys.modules.keys()):
    if 'semantic_matching' in module:
        del sys.modules[module]

from semantic_matching import ProductMatcher
import pandas as pd

print("Setting up the semantic search...")
print(f"Using src directory: {src_dir}")
```

```python
# Initialize the semantic search system
print("Loading sentence transformer model...")
matcher = ProductMatcher()

print("Loading product database...")
matcher.load_data()

print("Creating semantic embeddings...")
matcher.create_embeddings()

print("Ready.")
```

```python
print("SEMANTIC SEARCH DEMO")
print("=" * 50)
print("Regional terminology mapping: US -> UK")
print()

query = "gasoline"
matches = matcher.find_matches(query, top_k=3)

print(f"User searches: '{query}'")
for i, match in enumerate(matches, 1):
    print(f"  {i}. {match['product_name']} ({match['similarity']:.3f})")
```
